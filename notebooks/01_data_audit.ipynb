## Первичная подготовка данных

Приведение ключевых полей источников к единому формату, преобразование дат и проверка возможности объединения данных.

```python
player_df["appid"] = (
    player_df["URL"]
    .str.extract(r"/app/(\d+)")
    .astype("Int64")
)


games_df = games_df.reset_index()

if "AppID" in games_df.columns:
    games_df = games_df.rename(columns={"AppID": "appid"})
elif "index" in games_df.columns:
    games_df = games_df.rename(columns={"index": "appid"})

player_df["Date"] = pd.to_datetime(
    player_df["Date"],
    errors="coerce"
)

reviews_df["review_date"] = pd.to_datetime(
    reviews_df["unix_timestamp_created"],
    unit="s",
    errors="coerce"
)

display(player_df[["Game_Name", "appid", "Date"]].head())
display(games_df[["appid", "Name", "Release date"]].head())
display(reviews_df[["appid", "review_date", "voted_up"]].head())
```

```python
player_ids = set(player_df["appid"].dropna().astype(int))
games_ids = set(games_df["appid"].dropna().astype(int))
review_ids = set(reviews_df["appid"].dropna().astype(int))

print("Уникальных игр в PLAYER:", len(player_ids))
print("Есть в GAMES:", len(player_ids & games_ids))
print("Есть в REVIEWS:", len(player_ids & review_ids))
print("Есть одновременно во всех источниках:",
      len(player_ids & games_ids & review_ids))
```

## Анализ качества данных

Проверка исходных данных на наличие пропущенных значений, дубликатов и некорректных значений перед дальнейшей обработкой.

```python
def data_quality(df, name):
    print(f"\n{name}")

    print("Размер:", df.shape)
    print("Дубликатов:", df.duplicated().sum())

    missing = df.isna().sum()
    missing = missing[missing > 0].sort_values(ascending=False)

    print("\nПропуски:")
    if len(missing) == 0:
        print("Нет")
    else:
        print(missing)


data_quality(player_df, "PLAYER")
data_quality(games_df, "GAMES")
data_quality(reviews_df, "REVIEWS")
```

```python
def owners_to_numeric(value):
    if pd.isna(value):
        return None

    parts = str(value).replace(",", "").split("-")

    if len(parts) != 2:
        return None

    lower = float(parts[0].strip())
    upper = float(parts[1].strip())

    return (lower + upper) / 2


games_clean["Estimated owners"] = (
    games_clean["Estimated owners"]
    .apply(owners_to_numeric)
)
```

## Подготовка данных об отзывах

Отзывы агрегируются по игре и месяцу для формирования признаков, характеризующих пользовательскую активность и отношение игроков.

```python
reviews_clean["month"] = (
    reviews_clean["review_date"]
    .dt.to_period("M")
    .dt.to_timestamp()
)

reviews_clean["voted_up"] = (
    reviews_clean["voted_up"].astype(int)
)

print("Период отзывов:")
print(reviews_clean["review_date"].min())
print(reviews_clean["review_date"].max())

print("\nКоличество отзывов:", len(reviews_clean))
print("Количество игр:", reviews_clean["appid"].nunique())
```

```python
reviews_monthly = (
    reviews_clean
    .groupby(["appid", "month"])
    .agg(
        review_count=("review", "size"),
        positive_share=("voted_up", "mean"),
        avg_playtime=("playtime_forever", "mean"),
        avg_votes_up=("votes_up", "mean")
    )
    .reset_index()
)

display(reviews_monthly.head(10))

print("Размер агрегированной таблицы:", reviews_monthly.shape)
```

```python
player_clean["month"] = (
    player_clean["Date"]
    .dt.to_period("M")
    .dt.to_timestamp()
)
```

```python
display(
    player_clean[
        ["appid", "Game_Name", "month", "Avg_players", "Peak_Players"]
    ].head()
)
```

```python
dataset = player_clean.merge(
    reviews_monthly,
    on=["appid", "month"],
    how="left"
)

print("PLAYER до объединения:", len(player_clean))
print("После объединения:", len(dataset))

print(
    "Строк с данными об отзывах:",
    dataset["review_count"].notna().sum()
)

print(
    "Игр с данными об отзывах:",
    dataset.loc[
        dataset["review_count"].notna(),
        "appid"
    ].nunique()
)
```

```python
dataset = dataset.merge(
    games_clean,
    on="appid",
    how="left"
)

print("Строк после добавления GAMES:", len(dataset))
print("Уникальных игр:", dataset["appid"].nunique())
```

```python
dataset = dataset.sort_values(
    ["appid", "month"]
).reset_index(drop=True)
```

```python
dataset["target_next_month"] = (
    dataset
    .groupby("appid")["Avg_players"]
    .shift(-1)
)
```

```python
display(
    dataset[
        ["appid", "Game_Name", "month",
         "Avg_players", "target_next_month"]
    ].head(10)
)

print(
    "Строк без target:",
    dataset["target_next_month"].isna().sum()
)
```

```python
grouped = dataset.groupby("appid")

dataset["avg_players_lag_1"] = (
    grouped["Avg_players"].shift(1)
)

dataset["avg_players_lag_3"] = (
    grouped["Avg_players"].shift(3)
)

dataset["avg_players_lag_6"] = (
    grouped["Avg_players"].shift(6)
)
```

```python
dataset["avg_players_roll_3"] = (
    grouped["Avg_players"]
    .transform(
        lambda x: x.shift(1).rolling(3).mean()
    )
)

dataset["avg_players_roll_6"] = (
    grouped["Avg_players"]
    .transform(
        lambda x: x.shift(1).rolling(6).mean()
    )
)
```

```python
dataset["year"] = dataset["month"].dt.year
dataset["month_num"] = dataset["month"].dt.month
dataset["quarter"] = dataset["month"].dt.quarter
```

```python
dataset["game_age_months"] = (
    (dataset["month"].dt.year - dataset["Release date"].dt.year) * 12
    + dataset["month"].dt.month
    - dataset["Release date"].dt.month
)
```

```python
full_rows = []

for appid, group in player_clean.groupby("appid"):
    group = group.sort_values("month").copy()

    full_months = pd.date_range(
        start=group["month"].min(),
        end=group["month"].max(),
        freq="MS"
    )

    group = (
        group
        .set_index("month")
        .reindex(full_months)
        .rename_axis("month")
        .reset_index()
    )

    group["appid"] = appid

    full_rows.append(group)

player_regular = pd.concat(
    full_rows,
    ignore_index=True
)

print("Строк исходно:", len(player_clean))
print("Строк после восстановления календаря:", len(player_regular))
print("Добавлено пропущенных месяцев:",
      len(player_regular) - len(player_clean))

print(
    "Строк без Avg_players:",
    player_regular["Avg_players"].isna().sum()
)
```

```python
dataset = player_regular.merge(
    reviews_monthly,
    on=["appid", "month"],
    how="left"
)

dataset = dataset.merge(
    games_clean,
    on="appid",
    how="left"
)

dataset = dataset.sort_values(
    ["appid", "month"]
).reset_index(drop=True)

print("Размер:", dataset.shape)
print("Уникальных игр:", dataset["appid"].nunique())
print("Пропущенных Avg_players:", dataset["Avg_players"].isna().sum())
```

```python
grouped = dataset.groupby("appid")

dataset["target_next_month"] = (
    grouped["Avg_players"].shift(-1)
)

dataset["avg_players_lag_1"] = (
    grouped["Avg_players"].shift(1)
)

dataset["avg_players_lag_3"] = (
    grouped["Avg_players"].shift(3)
)

dataset["avg_players_lag_6"] = (
    grouped["Avg_players"].shift(6)
)

dataset["avg_players_roll_3"] = (
    grouped["Avg_players"]
    .transform(lambda x: x.shift(1).rolling(3).mean())
)

dataset["avg_players_roll_6"] = (
    grouped["Avg_players"]
    .transform(lambda x: x.shift(1).rolling(6).mean())
)

dataset["year"] = dataset["month"].dt.year
dataset["month_num"] = dataset["month"].dt.month
dataset["quarter"] = dataset["month"].dt.quarter

dataset["game_age_months"] = (
    (dataset["month"].dt.year - dataset["Release date"].dt.year) * 12
    + dataset["month"].dt.month
    - dataset["Release date"].dt.month
)
```
